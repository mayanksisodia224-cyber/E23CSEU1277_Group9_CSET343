```python
# LOGISTIC REGRESSION - BREAST CANCER WISCONSIN (DIAGNOSTIC)

# 1. IMPORT LIBRARIES

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully!")


# TASK 1: DATA ACQUISITION AND EXPLORATION

print("\n" + "="*70)
print("TASK 1: DATA ACQUISITION AND EXPLORATION")
print("="*70)


# 1. DOWNLOAD / LOAD DATASET

data = load_breast_cancer()

# Create DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Add target column
df["Outcome"] = data.target

print("\nDataset loaded successfully!")

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())


# 2. BASIC DATA INFORMATION

print("\n" + "-"*70)
print("DATASET INFORMATION")
print("-"*70)

print("\nData types:")
print(df.dtypes)

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# 3. TARGET VARIABLE

df["Outcome"] = df["Outcome"].astype(int)

print("\nTarget distribution:")
print(df["Outcome"].value_counts())

print("\nTarget meaning:")
print("0 = Malignant")
print("1 = Benign")


# 4. SUMMARY STATISTICS

print("\n" + "-"*70)
print("SUMMARY STATISTICS")
print("-"*70)

print("\nMean:")
display(df.mean(numeric_only=True))

print("\nMedian:")
display(df.median(numeric_only=True))

print("\nStandard Deviation:")
display(df.std(numeric_only=True))

print("\nComplete descriptive statistics:")
display(df.describe())


# 5. CHECK MISSING VALUES

print("\n" + "-"*70)
print("MISSING VALUE CHECK")
print("-"*70)

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal missing values:", missing_values.sum())

if missing_values.sum() == 0:
    print("No missing values found.")
else:
    print("Missing values found. They will be imputed.")


# 6. CHECK DUPLICATES

print("\n" + "-"*70)
print("DUPLICATE CHECK")
print("-"*70)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")


# 7. OUTLIER DETECTION USING IQR

print("\n" + "-"*70)
print("OUTLIER DETECTION")
print("-"*70)

# Only feature columns
feature_columns = data.feature_names.tolist()

outlier_summary = {}

for column in feature_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary[column] = len(outliers)

outlier_df = pd.DataFrame(
    list(outlier_summary.items()),
    columns=["Feature", "Number_of_Outliers"]
)

display(outlier_df.sort_values(
    "Number_of_Outliers",
    ascending=False
).head(15))


# 8. HANDLE OUTLIERS

df_clean = df.copy()

for column in feature_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean[column] = df_clean[column].clip(
        lower_bound,
        upper_bound
    )

print("\nOutlier handling completed using IQR clipping.")


# 9. CHECK MISSING VALUES AGAIN

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum().sum())


# VISUALIZATION

print("\n" + "="*70)
print("DATA VISUALIZATION")
print("="*70)


# 10. CLASS DISTRIBUTION

plt.figure(figsize=(7, 5))

sns.countplot(
    x="Outcome",
    data=df_clean
)

plt.title("Class Distribution")
plt.xlabel("Outcome (0 = Malignant, 1 = Benign)")
plt.ylabel("Number of Samples")

plt.show()


# Print percentages
class_counts = df_clean["Outcome"].value_counts()
class_percentages = df_clean["Outcome"].value_counts(normalize=True) * 100

print("\nClass Counts:")
print(class_counts)

print("\nClass Percentages:")
print(class_percentages.round(2))


# 11. HISTOGRAMS OF KEY FEATURES

key_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness"
]

for feature in key_features:

    plt.figure(figsize=(7, 5))

    sns.histplot(
        data=df_clean,
        x=feature,
        hue="Outcome",
        kde=True
    )

    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")

    plt.show()


# 12. BOX PLOTS

plt.figure(figsize=(14, 6))

sns.boxplot(
    data=df_clean[key_features]
)

plt.title("Box Plot of Key Features")
plt.xticks(rotation=30)

plt.show()



# 13. CORRELATION HEATMAP

plt.figure(figsize=(18, 14))

correlation_matrix = df_clean[feature_columns].corr()

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Features")

plt.show()


# 14. TOP CORRELATED FEATURES

corr_pairs = correlation_matrix.abs().unstack()

corr_pairs = corr_pairs[
    corr_pairs < 1
].sort_values(
    ascending=False
)

print("\nTop correlated feature pairs:")
display(corr_pairs.head(10))


# CLASS IMBALANCE DISCUSSION

print("\n" + "="*70)
print("CLASS DISTRIBUTION ANALYSIS")
print("="*70)

print("""
The dataset contains two classes:

0 = Malignant
1 = Benign

The classes are not perfectly balanced. There are more benign
samples than malignant samples.

This imbalance can affect model training because a model may
become biased toward the majority class.

For example, a model could obtain relatively high accuracy by
predicting the majority class more frequently.

Therefore, accuracy alone should not be used to judge a medical
diagnosis model.

Precision, Recall, F1-score, Confusion Matrix and ROC-AUC are
also important.

Stratified train-test splitting is used below so that the
proportion of malignant and benign cases is maintained in both
training and testing datasets.
""")


# TASK 2: DATA PREPROCESSING

print("\n" + "="*70)
print("TASK 2: DATA PREPROCESSING")
print("="*70)


# 15. SEPARATE FEATURES AND TARGET

X = df_clean[feature_columns]

y = df_clean["Outcome"]

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)


# 16. HANDLE MISSING VALUES

imputer = SimpleImputer(strategy="median")

X_imputed = imputer.fit_transform(X)

print("\nMissing values handled using median imputation.")


# 17. TRAIN-TEST SPLIT

X_train, X_test, y_train, y_test = train_test_split(
    X_imputed,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])



# 18. CHECK STRATIFIED DISTRIBUTION

print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).round(3))



# 19. STANDARDIZATION

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\nFeature standardization completed.")

print("\nExample of scaled training data:")
print(X_train_scaled[:3])



# TASK 3: MODEL TRAINING AND BASIC EVALUATION

print("\n" + "="*70)
print("TASK 3: LOGISTIC REGRESSION MODEL")
print("="*70)



# 20. TRAIN LOGISTIC REGRESSION

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model.fit(
    X_train_scaled,
    y_train
)

print("\nLogistic Regression model trained successfully!")


# 21. MAKE PREDICTIONS

y_pred = model.predict(X_test_scaled)

# Probability of class 1
y_probability = model.predict_proba(X_test_scaled)[:, 1]



# MODEL EVALUATION

# 22. ACCURACY

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n" + "-"*70)
print("ACCURACY")
print("-"*70)

print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy: {accuracy*100:.2f}%")


# 23. PRECISION

precision = precision_score(
    y_test,
    y_pred,
    average=None
)

print("\n" + "-"*70)
print("PRECISION FOR BOTH CLASSES")
print("-"*70)

print(f"Class 0 (Malignant): {precision[0]:.4f}")
print(f"Class 1 (Benign):   {precision[1]:.4f}")


# 24. RECALL

recall = recall_score(
    y_test,
    y_pred,
    average=None
)

print("\n" + "-"*70)
print("RECALL FOR BOTH CLASSES")
print("-"*70)

print(f"Class 0 (Malignant): {recall[0]:.4f}")
print(f"Class 1 (Benign):   {recall[1]:.4f}")


# 25. F1 SCORE

f1 = f1_score(
    y_test,
    y_pred,
    average=None
)

print("\n" + "-"*70)
print("F1-SCORE FOR BOTH CLASSES")
print("-"*70)

print(f"Class 0 (Malignant): {f1[0]:.4f}")
print(f"Class 1 (Benign):   {f1[1]:.4f}")


# 26. COMPLETE CLASSIFICATION REPORT

print("\n" + "-"*70)
print("CLASSIFICATION REPORT")
print("-"*70)

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Malignant", "Benign"]
    )
)


# 27. CONFUSION MATRIX

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\n" + "-"*70)
print("CONFUSION MATRIX")
print("-"*70)

print(cm)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Malignant", "Benign"],
    yticklabels=["Malignant", "Benign"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()


# 28. ROC CURVE AND AUC

fpr, tpr, thresholds = roc_curve(
    y_test,
    y_probability
)

auc_score = roc_auc_score(
    y_test,
    y_probability
)

print("\n" + "-"*70)
print("ROC-AUC")
print("-"*70)

print(f"AUC Score: {auc_score:.4f}")


plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {auc_score:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve - Logistic Regression")

plt.legend()

plt.grid()

plt.show()


# 29. MODEL COEFFICIENTS / INTERPRETATION

print("\n" + "="*70)
print("MODEL INTERPRETATION")
print("="*70)

coefficients = pd.DataFrame({
    "Feature": feature_columns,
    "Coefficient": model.coef_[0],
    "Absolute_Coefficient": np.abs(model.coef_[0])
})

coefficients = coefficients.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

print("\nMost influential features:")
display(coefficients.head(15))


# 30. FEATURE IMPORTANCE VISUALIZATION

top_features = coefficients.head(10).sort_values(
    "Coefficient"
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["Feature"],
    top_features["Coefficient"]
)

plt.xlabel("Logistic Regression Coefficient")
plt.ylabel("Feature")

plt.title(
    "Top 10 Features - Logistic Regression Coefficients"
)

plt.axvline(
    x=0,
    linestyle="--"
)

plt.show()


# 31. FINAL RESULTS SUMMARY

print("\n" + "="*70)
print("FINAL MODEL RESULTS")
print("="*70)

results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision - Malignant",
        "Recall - Malignant",
        "F1 - Malignant",
        "Precision - Benign",
        "Recall - Benign",
        "F1 - Benign",
        "ROC-AUC"
    ],
    "Score": [
        accuracy,
        precision[0],
        recall[0],
        f1[0],
        precision[1],
        recall[1],
        f1[1],
        auc_score
    ]
})

display(results)

```
