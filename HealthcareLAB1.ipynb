```python

import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler


pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")



# 1. DATASET ACQUISITION

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00519/heart_failure_clinical_records_dataset.csv"

try:
    df = pd.read_csv(url)
    print("Dataset downloaded successfully!")
except Exception as e:
    print("Automatic download failed.")
    print("Please upload 'heart_failure_clinical_records_dataset.csv' to Colab.")

    from google.colab import files
    uploaded = files.upload()

    filename = list(uploaded.keys())[0]
    df = pd.read_csv(filename)

print("\nDataset shape:", df.shape)



# 2. LOAD DATASET - FIRST 5 ROWS

print("\nFIRST 5 ROWS\n")
display(df.head())



# 3. SUMMARIZE DATASET

print("\nDATASET SUMMARY\n")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDataset Information:")
df.info()



# 4. VISUALIZE MISSING VALUES

plt.figure(figsize=(14, 7))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Heatmap of Missing Values")
plt.xlabel("Features")
plt.ylabel("Records")
plt.show()


# 5. PLOT DISTRIBUTIONS OF NUMERICAL FEATURES

# Select numerical columns
numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("\nNumerical Features:")
print(numerical_columns)


# Histograms
df[numerical_columns].hist(
    figsize=(16, 14),
    bins=20
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()


# Boxplots
for column in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)

    plt.show()



# 6. SAVE ORIGINAL DATA FOR COMPARISON

df_before_cleaning = df.copy()

print("\nOriginal dataset shape:", df.shape)



# 7. CORRECT INCONSISTENCIES

print("\nCHECKING INCONSISTENCIES\n")

# Check negative values in columns where they are invalid

# Age
negative_age = (df["age"] < 0).sum()
print("Negative age values:", negative_age)

# Ejection fraction
negative_ef = (df["ejection_fraction"] < 0).sum()
print("Negative ejection fraction values:", negative_ef)

# Serum creatinine
negative_creatinine = (df["serum_creatinine"] < 0).sum()
print("Negative serum creatinine values:", negative_creatinine)

# Platelets
negative_platelets = (df["platelets"] < 0).sum()
print("Negative platelet values:", negative_platelets)



# Fix invalid values

# Replace invalid age values with NaN
df.loc[df["age"] < 0, "age"] = np.nan

# Replace invalid ejection fraction
df.loc[df["ejection_fraction"] < 0, "ejection_fraction"] = np.nan

# Replace invalid serum creatinine
df.loc[df["serum_creatinine"] < 0, "serum_creatinine"] = np.nan

# Replace invalid platelets
df.loc[df["platelets"] < 0, "platelets"] = np.nan

print("\nInvalid numerical values corrected.")



# 8. IMPUTE MISSING VALUES

print("\nMISSING VALUES BEFORE IMPUTATION\n")

print(df.isnull().sum())


# Numerical Imputation

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

for column in numerical_columns:

    if df[column].isnull().sum() > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(median_value)

        print(
            f"{column}: Missing values replaced with median = {median_value}"
        )



# 9. CATEGORICAL FEATURES

categorical_columns = [
    "anaemia",
    "diabetes",
    "high_blood_pressure",
    "sex",
    "smoking"
]

print("\nCategorical Columns:")
print(categorical_columns)



# Mode Imputation

for column in categorical_columns:

    if column in df.columns:

        if df[column].isnull().sum() > 0:

            mode_value = df[column].mode()[0]

            df[column] = df[column].fillna(mode_value)

            print(
                f"{column}: Missing values replaced with mode = {mode_value}"
            )



# 10. REMOVE OUTLIERS

print("\nOUTLIER REMOVAL\n")

def remove_outliers_iqr(data, columns):

    cleaned_data = data.copy()

    for column in columns:

        Q1 = cleaned_data[column].quantile(0.25)
        Q3 = cleaned_data[column].quantile(0.75)

        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        before = len(cleaned_data)

        cleaned_data = cleaned_data[
            (cleaned_data[column] >= lower_bound) &
            (cleaned_data[column] <= upper_bound)
        ]

        after = len(cleaned_data)

        print(
            f"{column}: Removed {before - after} outliers"
        )

    return cleaned_data


# Only apply IQR to continuous variables.
outlier_columns = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]

df = remove_outliers_iqr(
    df,
    outlier_columns
)

print("\nDataset shape after outlier removal:", df.shape)



# 11. FEATURE ENGINEERING - AGE GROUP

print("\nFEATURE ENGINEERING\n")

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 40, 60, np.inf],
    labels=["<40", "40-60", ">60"],
    right=False
)

print("Age group created successfully!")

print("\nAge Group Distribution:")
print(df["age_group"].value_counts())



# 12. FEATURE ENGINEERING - RISK SCORE


df["risk_score_raw"] = (
    df["ejection_fraction"] *
    df["serum_creatinine"]
)

# Normalize risk score between 0 and 1

risk_scaler = MinMaxScaler()

df["risk_score"] = risk_scaler.fit_transform(
    df[["risk_score_raw"]]
)

# Remove temporary column
df.drop(
    columns=["risk_score_raw"],
    inplace=True
)

print("\nRisk score created successfully!")

print(df[[
    "ejection_fraction",
    "serum_creatinine",
    "risk_score"
]].head())



# 13. ENCODE CATEGORICAL VARIABLES


df = pd.get_dummies(
    df,
    columns=["age_group"],
    prefix="age_group",
    dtype=int
)

print("\nCategorical encoding completed!")

print("\nColumns after encoding:")
print(df.columns.tolist())



# 14. NORMALIZE NUMERICAL FEATURES

print("\nNORMALIZATION\n")

# Numerical continuous features
scale_columns = [
    "age",
    "creatinine_phosphokinase",
    "ejection_fraction",
    "platelets",
    "serum_creatinine",
    "serum_sodium",
    "time"
]

# Make sure all selected columns exist
scale_columns = [
    col for col in scale_columns
    if col in df.columns
]

scaler = MinMaxScaler()

df[scale_columns] = scaler.fit_transform(
    df[scale_columns]
)

print("Numerical features normalized using MinMaxScaler.")

print("\nNormalized Data:")
display(df[scale_columns].head())



# 15. COMPUTE SUMMARY STATISTICS - BEFORE CLEANING

print("\nSTATISTICS BEFORE CLEANING\n")

stats_before = df_before_cleaning[
    numerical_columns
].agg(
    ["mean", "median", "std"]
).T

display(stats_before)



# 16. COMPUTE SUMMARY STATISTICS - AFTER CLEANING

print("\nSTATISTICS AFTER CLEANING\n")

# Some numerical columns still exist after cleaning
available_numeric = [
    col for col in numerical_columns
    if col in df.columns
]

stats_after = df[
    available_numeric
].agg(
    ["mean", "median", "std"]
).T

display(stats_after)


# 17. COMPARE BEFORE AND AFTER CLEANING


print("\nSTATISTICS COMPARISON\n")

comparison = pd.DataFrame({
    "Mean Before": df_before_cleaning[
        available_numeric
    ].mean(),

    "Mean After": df[
        available_numeric
    ].mean(),

    "Median Before": df_before_cleaning[
        available_numeric
    ].median(),

    "Median After": df[
        available_numeric
    ].median(),

    "Std Before": df_before_cleaning[
        available_numeric
    ].std(),

    "Std After": df[
        available_numeric
    ].std()
})

display(comparison)



# 18. VALIDATE CLEANING - HISTOGRAMS


print("\nCLEANED DATA DISTRIBUTIONS\n")

df[available_numeric].hist(
    figsize=(16, 14),
    bins=20
)

plt.suptitle(
    "Distributions After Data Cleaning",
    fontsize=16
)

plt.tight_layout()
plt.show()



# 19. VALIDATE CLEANING - BOXPLOTS


for column in available_numeric:

    plt.figure(figsize=(8, 4))

    sns.boxplot(
        x=df[column]
    )

    plt.title(
        f"Cleaned Boxplot - {column}"
    )

    plt.xlabel(column)

    plt.show()



# 20. CHECK MISSING VALUES AFTER CLEANING


print("\nFINAL MISSING VALUE CHECK\n")

missing_after = df.isnull().sum()

print(missing_after)

print(
    "\nTotal missing values after cleaning:",
    missing_after.sum()
)

if missing_after.sum() == 0:
    print("SUCCESS: No missing values remain!")
else:
    print("Some missing values remain. Review the columns above.")



# 21. CHECK DUPLICATES


print("\nDUPLICATE CHECK\n")

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)

if duplicates > 0:

    df = df.drop_duplicates()

    print(
        "Duplicates removed."
    )

else:

    print(
        "No duplicate rows found."
    )



# 22. FINAL DATASET INFORMATION


print("\nFINAL DATASET\n")

print("Final number of rows:", df.shape[0])
print("Final number of columns:", df.shape[1])

print("\nFinal columns:")
print(df.columns.tolist())

print("\nFinal data types:")
print(df.dtypes)

print("\nFinal first 5 rows:")
display(df.head())



# 23. SAVE CLEANED DATASET


output_file = "heart_failure_cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    f"\nCleaned dataset saved as: {output_file}"
)



# 24. DOWNLOAD CLEANED DATASET


try:

    from google.colab import files

    files.download(
        output_file
    )

except:

    print(
        "If running outside Google Colab, "
        "the file is available in the current directory."
    )



# END OF LAB ASSIGNMENT

print("LAB ASSIGNMENT - 2 COMPLETED SUCCESSFULLY")
```
