```python
# 0. INSTALL REQUIRED LIBRARIES

!pip -q install kagglehub wfdb pydicom imbalanced-learn nltk


# 1. IMPORT LIBRARIES

import os
import re
import glob
import shutil
import warnings
import zipfile
import random

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image, ImageOps

import scipy.signal as signal
from scipy import stats

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    LabelEncoder
)

from sklearn.feature_selection import SelectKBest, f_classif

from sklearn.decomposition import PCA

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

from imblearn.over_sampling import RandomOverSampler

import kagglehub
import wfdb

warnings.filterwarnings("ignore")

np.random.seed(42)
random.seed(42)

print("All libraries imported successfully!")



# CREATE WORKING DIRECTORIES

BASE_DIR = "/content/medical_multimodal_lab"

TABULAR_DIR = os.path.join(BASE_DIR, "tabular")
TEXT_DIR = os.path.join(BASE_DIR, "text")
IMAGE_DIR = os.path.join(BASE_DIR, "image")
SIGNAL_DIR = os.path.join(BASE_DIR, "signal")

for folder in [
    BASE_DIR,
    TABULAR_DIR,
    TEXT_DIR,
    IMAGE_DIR,
    SIGNAL_DIR
]:
    os.makedirs(folder, exist_ok=True)

print("Working directories created:")
print(BASE_DIR)


# PART A - TABULAR DATA
# PIMA INDIANS DIABETES DATASET

print("\n")
print("=" * 70)
print("PART A - TABULAR DATA")
print("=" * 70)



# A1. DOWNLOAD DATASET


print("\nDownloading Pima Indians Diabetes Dataset...")

try:

    pima_path = kagglehub.dataset_download(
        "uciml/pima-indians-diabetes-database"
    )

    print("Dataset downloaded to:")
    print(pima_path)

except Exception as e:

    print("Kaggle download failed.")
    print("Trying direct public CSV fallback...")

    fallback_url = (
        "https://raw.githubusercontent.com/jbrownlee/"
        "Datasets/master/pima-indians-diabetes.data.csv"
    )

    df = pd.read_csv(
        fallback_url,
        header=None
    )

    df.columns = [
        "Pregnancies",
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI",
        "DiabetesPedigreeFunction",
        "Age",
        "Outcome"
    ]

    print("Fallback dataset downloaded successfully.")


# A2. LOCATE CSV

if "df" not in locals():

    csv_files = glob.glob(
        os.path.join(pima_path, "**", "*.csv"),
        recursive=True
    )

    print("CSV files found:")
    for f in csv_files:
        print(f)

    if len(csv_files) == 0:
        raise FileNotFoundError(
            "No CSV file found in downloaded dataset."
        )

    df = pd.read_csv(csv_files[0])


# A3. INSPECT DATASET

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()



# A4. REMOVE DUPLICATES

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print(
    "Shape after removing duplicates:",
    df.shape
)


# A5. HEALTHCARE-SPECIFIC MISSING VALUES

clinical_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for col in clinical_zero_columns:
    df[col] = df[col].replace(0, np.nan)


print("\nMissing values after converting invalid zeros:")

missing_table = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(missing_table)



# A6. IMPUTE MISSING VALUES


for col in clinical_zero_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )


print("\nMissing values after median imputation:")
print(df.isnull().sum())



# A7. CLINICAL RANGE VALIDATION

print("\nClinical plausibility checks:")

checks = {

    "Glucose < 0":
        (df["Glucose"] < 0).sum(),

    "BloodPressure < 0":
        (df["BloodPressure"] < 0).sum(),

    "BMI < 0":
        (df["BMI"] < 0).sum(),

    "Age < 0":
        (df["Age"] < 0).sum(),

    "Pregnancies < 0":
        (df["Pregnancies"] < 0).sum()
}

for name, count in checks.items():
    print(f"{name}: {count}")



# A8. FEATURE ENGINEERING

df["BMI_Category"] = pd.cut(
    df["BMI"],
    bins=[0, 18.5, 25, 30, 100],
    labels=[
        "Underweight",
        "Normal",
        "Overweight",
        "Obese"
    ]
)

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[20, 30, 40, 50, 100],
    labels=[
        "21-30",
        "31-40",
        "41-50",
        "51+"
    ]
)

df["Glucose_Category"] = pd.cut(
    df["Glucose"],
    bins=[0, 99, 125, 1000],
    labels=[
        "Normal",
        "Prediabetes",
        "High"
    ]
)

print("\nFeature engineering completed.")

display(
    df[
        [
            "Glucose",
            "BMI",
            "Age",
            "BMI_Category",
            "Age_Group",
            "Glucose_Category",
            "Outcome"
        ]
    ].head()
)



# A9. TABULAR EDA

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Outcome"
)

plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Count")
plt.show()



# A10. DISTRIBUTION OF NUMERIC VARIABLES

numeric_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

df[numeric_columns].hist(
    figsize=(15, 12),
    bins=20
)

plt.suptitle(
    "Distribution of Medical Variables",
    fontsize=16
)

plt.tight_layout()
plt.show()


# A11. CORRELATION HEATMAP

plt.figure(figsize=(12, 8))

sns.heatmap(
    df[numeric_columns + ["Outcome"]].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()



# A12. OUTLIER DETECTION

plt.figure(figsize=(15, 6))

sns.boxplot(
    data=df[numeric_columns]
)

plt.xticks(
    rotation=45
)

plt.title("Boxplot for Outlier Detection")
plt.show()



# A13. HYPOTHESIS TEST


diabetic_glucose = df[
    df["Outcome"] == 1
]["Glucose"]

non_diabetic_glucose = df[
    df["Outcome"] == 0
]["Glucose"]

t_stat, p_value = stats.ttest_ind(
    diabetic_glucose,
    non_diabetic_glucose,
    equal_var=False
)

print("\n" + "=" * 60)
print("HYPOTHESIS TEST - T-TEST")
print("=" * 60)

print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print(
        "Conclusion: Reject H0."
    )
    print(
        "There is a statistically significant difference "
        "in glucose levels between the two groups."
    )
else:
    print(
        "Conclusion: Fail to reject H0."
    )



# A14. CHI-SQUARE TEST

#
# Test relationship between BMI category and diabetes outcome.
#
# H0: BMI category and diabetes outcome are independent.
# H1: They are associated.


contingency_table = pd.crosstab(
    df["BMI_Category"],
    df["Outcome"]
)

print("\nContingency Table:")
display(contingency_table)

chi2, p, dof, expected = stats.chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)

if p < 0.05:
    print(
        "Conclusion: Reject H0."
    )
    print(
        "BMI category and diabetes outcome "
        "are statistically associated."
    )
else:
    print(
        "Conclusion: Fail to reject H0."
    )



# A15. ANOVA TEST

anova_groups = []

for group in df["Age_Group"].dropna().unique():

    values = df[
        df["Age_Group"] == group
    ]["Glucose"]

    anova_groups.append(values)


f_stat, anova_p = stats.f_oneway(
    *anova_groups
)

print("\n" + "=" * 60)
print("ONE-WAY ANOVA")
print("=" * 60)

print("F-statistic:", f_stat)
print("P-value:", anova_p)

if anova_p < 0.05:
    print(
        "Conclusion: Reject H0."
    )
    print(
        "At least one age group has a significantly "
        "different mean glucose level."
    )
else:
    print(
        "Conclusion: Fail to reject H0."
    )



# A16. FEATURE SELECTION

X = df[numeric_columns]
y = df["Outcome"]

selector = SelectKBest(
    score_func=f_classif,
    k=5
)

X_selected = selector.fit_transform(
    X,
    y
)

selected_features = X.columns[
    selector.get_support()
]

print("\nSelected features:")
print(selected_features.tolist())



# A17. FEATURE SCALING

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=numeric_columns
)

print("\nScaled features:")
display(X_scaled.head())



# A18. CLASS IMBALANCE

print("\nOriginal class distribution:")
print(y.value_counts())

plt.figure(figsize=(6, 4))

sns.countplot(
    x=y
)

plt.title("Class Distribution")
plt.show()



# A19. HANDLE CLASS IMBALANCE

ros = RandomOverSampler(
    random_state=42
)

X_balanced, y_balanced = ros.fit_resample(
    X_scaled,
    y
)

print("\nBalanced class distribution:")
print(
    pd.Series(y_balanced).value_counts()
)



# PART B - TEXTUAL DATA
# MTSAMPLES MEDICAL TRANSCRIPTIONS

print("\n")
print("=" * 70)
print("PART B - TEXTUAL DATA")
print("=" * 70)



# B1. DOWNLOAD MTSAMPLES

print("\nDownloading MTSamples dataset...")

try:

    mtsamples_path = kagglehub.dataset_download(
        "tboyle10/medicaltranscriptions"
    )

    print(
        "MTSamples downloaded to:",
        mtsamples_path
    )

except Exception as e:

    print("\nCould not download MTSamples automatically.")
    print("Error:")
    print(e)

    print(
        "\nIf Kaggle requests authentication, "
        "configure Kaggle credentials in Colab and "
        "rerun this section."
    )

    mtsamples_path = None



# B2. LOCATE TEXT CSV


if mtsamples_path is not None:

    text_csv_files = glob.glob(
        os.path.join(
            mtsamples_path,
            "**",
            "*.csv"
        ),
        recursive=True
    )

    print("\nCSV files found:")

    for f in text_csv_files:
        print(f)

    if len(text_csv_files) > 0:

        text_file = text_csv_files[0]

        text_df = pd.read_csv(
            text_file
        )

        print(
            "\nLoaded file:",
            text_file
        )

    else:

        print(
            "No CSV file found."
        )

        text_df = None

else:

    text_df = None



# B3. INSPECT TEXT DATA

if text_df is not None:

    print("\nFirst 5 records:")
    display(text_df.head())

    print("\nShape:")
    print(text_df.shape)

    print("\nColumns:")
    print(text_df.columns.tolist())

    print("\nMissing values:")
    print(text_df.isnull().sum())



# B4. REMOVE DUPLICATES


if text_df is not None:

    print(
        "\nDuplicate records:",
        text_df.duplicated().sum()
    )

    text_df = text_df.drop_duplicates()



# B5. IDENTIFY TRANSCRIPTION COLUMN


if text_df is not None:

    possible_text_columns = [
        "transcription",
        "Transcription",
        "text",
        "TEXT"
    ]

    transcription_column = None

    for col in possible_text_columns:

        if col in text_df.columns:
            transcription_column = col
            break

    if transcription_column is None:

        # Find the longest text-like column
        object_columns = text_df.select_dtypes(
            include=["object"]
        ).columns

        if len(object_columns) > 0:

            avg_lengths = {}

            for col in object_columns:

                avg_lengths[col] = (
                    text_df[col]
                    .fillna("")
                    .astype(str)
                    .str.len()
                    .mean()
                )

            transcription_column = max(
                avg_lengths,
                key=avg_lengths.get
            )

    print(
        "\nSelected transcription column:",
        transcription_column
    )


# B6. CLEAN MEDICAL TEXT


def clean_medical_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove email addresses
    text = re.sub(
        r"\b[\w\.-]+@[\w\.-]+\.\w+\b",
        " ",
        text
    )

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+",
        " ",
        text
    )

    # Remove phone numbers
    text = re.sub(
        r"\b\d{3}[-.\s]?\d{3}[-.\s]?\d{4}\b",
        " ",
        text
    )

    # Remove dates
    text = re.sub(
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",
        " ",
        text
    )

    # Remove potential patient IDs
    text = re.sub(
        r"\b(patient|mrn|medical record|record id)"
        r"\s*[:#]?\s*\w+\b",
        " ",
        text,
        flags=re.IGNORECASE
    )

    # Remove non-alphabetic characters
    text = re.sub(
        r"[^a-zA-Z\s]",
        " ",
        text
    )

    # Remove extra whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


if text_df is not None:

    text_df["clean_transcription"] = (
        text_df[transcription_column]
        .apply(clean_medical_text)
    )

    print("\nCleaned text:")
    display(
        text_df[
            [
                transcription_column,
                "clean_transcription"
            ]
        ].head()
    )



# B7. TEXT LENGTH FEATURE ENGINEERING

if text_df is not None:

    text_df["word_count"] = (
        text_df["clean_transcription"]
        .str.split()
        .str.len()
    )

    text_df["character_count"] = (
        text_df["clean_transcription"]
        .str.len()
    )

    text_df["sentence_count"] = (
        text_df["clean_transcription"]
        .str.count(r"[.!?]")
    )

    print("\nText features:")
    display(
        text_df[
            [
                "word_count",
                "character_count",
                "sentence_count"
            ]
        ].describe()
    )



# B8. TEXT EDA


if text_df is not None:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        text_df["word_count"],
        bins=40
    )

    plt.title(
        "Distribution of Medical Transcription Length"
    )

    plt.xlabel("Number of Words")
    plt.ylabel("Number of Transcriptions")

    plt.show()



# B9. MEDICAL SPECIALTY DISTRIBUTION


if text_df is not None:

    specialty_column = None

    for col in [
        "medical_specialty",
        "Medical Specialty",
        "specialty"
    ]:

        if col in text_df.columns:
            specialty_column = col
            break

    if specialty_column is not None:

        print("\nTop Medical Specialties:")

        display(
            text_df[
                specialty_column
            ]
            .value_counts()
            .head(15)
        )

        plt.figure(figsize=(12, 6))

        text_df[
            specialty_column
        ].value_counts().head(15).plot(
            kind="bar"
        )

        plt.title(
            "Top Medical Specialties"
        )

        plt.xlabel(
            "Medical Specialty"
        )

        plt.ylabel(
            "Number of Records"
        )

        plt.xticks(rotation=60)

        plt.show()



# B10. SIMPLE TOKENIZATION


if text_df is not None:

    def tokenize(text):

        return text.split()

    text_df["tokens"] = (
        text_df["clean_transcription"]
        .apply(tokenize)
    )

    print("\nExample token sequence:")

    if len(text_df) > 0:
        print(
            text_df["tokens"].iloc[0][:30]
        )


# B11. VOCABULARY CREATION


if text_df is not None:

    from collections import Counter

    counter = Counter()

    for tokens in text_df["tokens"]:
        counter.update(tokens)

    print(
        "\nVocabulary size:",
        len(counter)
    )

    print(
        "\nMost common medical words:"
    )

    print(
        counter.most_common(20)
    )



# B12. TEXT TO INTEGER SEQUENCES


if text_df is not None:

    MAX_VOCAB = 10000
    MAX_LENGTH = 100

    most_common_words = [
        word
        for word, count in
        counter.most_common(MAX_VOCAB - 2)
    ]

    word_to_id = {
        "<PAD>": 0,
        "<UNK>": 1
    }

    for idx, word in enumerate(
        most_common_words,
        start=2
    ):
        word_to_id[word] = idx

    def text_to_sequence(text):

        tokens = text.split()

        sequence = [
            word_to_id.get(
                token,
                word_to_id["<UNK>"]
            )
            for token in tokens
        ]

        sequence = sequence[:MAX_LENGTH]

        sequence += [
            word_to_id["<PAD>"]
        ] * (
            MAX_LENGTH - len(sequence)
        )

        return sequence

    text_df["token_sequence"] = (
        text_df["clean_transcription"]
        .apply(text_to_sequence)
    )

    token_matrix = np.array(
        text_df["token_sequence"].tolist()
    )

    print(
        "\nToken matrix shape:",
        token_matrix.shape
    )

    print(
        "Example sequence:",
        token_matrix[0][:30]
    )



# PART C - IMAGE DATA
# CHEST X-RAY PNEUMONIA

print("\n")
print("=" * 70)
print("PART C - IMAGE DATA")
print("=" * 70)




# C1. DOWNLOAD CHEST X-RAY DATASET


print(
    "\nDownloading Chest X-Ray Pneumonia Dataset..."
)

try:

    chest_path = kagglehub.dataset_download(
        "paultimothymooney/chest-xray-pneumonia"
    )

    print(
        "Dataset downloaded to:",
        chest_path
    )

except Exception as e:

    print("\nChest X-Ray download failed.")

    print(e)

    chest_path = None



# C2. LOCATE IMAGE DIRECTORIES


if chest_path is not None:

    print("\nDataset directory structure:")

    for root, dirs, files in os.walk(chest_path):

        level = root.replace(
            chest_path,
            ""
        ).count(os.sep)

        if level <= 2:

            print(
                "  " * level +
                os.path.basename(root) +
                "/"
            )



# C3. FIND CHEST X-RAY IMAGES


if chest_path is not None:

    image_files = glob.glob(
        os.path.join(
            chest_path,
            "**",
            "*.jpeg"
        ),
        recursive=True
    )

    image_files += glob.glob(
        os.path.join(
            chest_path,
            "**",
            "*.jpg"
        ),
        recursive=True
    )

    image_files += glob.glob(
        os.path.join(
            chest_path,
            "**",
            "*.png"
        ),
        recursive=True
    )

    print(
        "\nTotal images found:",
        len(image_files)
    )



# C4. IMAGE CLASS DISTRIBUTION


if chest_path is not None:

    image_labels = []

    for file in image_files:

        file_lower = file.lower()

        if "pneumonia" in file_lower:
            image_labels.append("PNEUMONIA")

        elif "normal" in file_lower:
            image_labels.append("NORMAL")

        else:
            image_labels.append("UNKNOWN")

    image_label_df = pd.DataFrame({
        "file": image_files,
        "label": image_labels
    })

    print("\nImage class distribution:")

    display(
        image_label_df[
            "label"
        ].value_counts()
    )



# C5. IMAGE EDA - DISPLAY SAMPLE IMAGES


if chest_path is not None:

    sample_df = image_label_df[
        image_label_df["label"] != "UNKNOWN"
    ].sample(
        min(
            8,
            len(
                image_label_df[
                    image_label_df["label"] != "UNKNOWN"
                ]
            )
        ),
        random_state=42
    )

    plt.figure(figsize=(14, 8))

    for i, (_, row) in enumerate(
        sample_df.iterrows()
    ):

        img = Image.open(
            row["file"]
        ).convert("L")

        plt.subplot(
            2,
            4,
            i + 1
        )

        plt.imshow(
            img,
            cmap="gray"
        )

        plt.title(
            row["label"]
        )

        plt.axis("off")

    plt.suptitle(
        "Sample Chest X-Ray Images"
    )

    plt.tight_layout()
    plt.show()



# C6. IMAGE SIZE ANALYSIS

if chest_path is not None:

    sizes = []

    for file in sample_df["file"]:

        try:

            img = Image.open(file)

            sizes.append(
                img.size
            )

        except:
            pass

    print(
        "\nSample image sizes:"
    )

    print(sizes)



# C7. IMAGE METADATA / PHI CHECK

if chest_path is not None:

    print(
        "\nChecking image metadata..."
    )

    if len(image_files) > 0:

        test_image = Image.open(
            image_files[0]
        )

        metadata = test_image.getexif()

        print(
            "Number of EXIF metadata fields:",
            len(metadata)
        )

        if len(metadata) > 0:

            print(
                "Metadata detected."
            )

            for key, value in metadata.items():
                print(
                    key,
                    ":",
                    value
                )

        else:

            print(
                "No EXIF metadata detected."
            )



# C8. IMAGE PREPROCESSING

IMG_SIZE = (224, 224)

def preprocess_xray(image_path):

    img = Image.open(
        image_path
    ).convert("L")

    # Resize
    img = img.resize(
        IMG_SIZE
    )

    # Convert to numpy
    arr = np.array(
        img,
        dtype=np.float32
    )

    # Normalize to [0,1]
    arr = arr / 255.0

    # Add channel dimension
    arr = np.expand_dims(
        arr,
        axis=-1
    )

    return arr


if chest_path is not None:

    valid_images = image_label_df[
        image_label_df["label"] != "UNKNOWN"
    ].sample(
        min(20, len(image_label_df)),
        random_state=42
    )

    image_tensor_list = []

    for file in valid_images["file"]:

        try:

            arr = preprocess_xray(
                file
            )

            image_tensor_list.append(
                arr
            )

        except Exception as e:

            print(
                "Image processing error:",
                e
            )

    if len(image_tensor_list) > 0:

        image_tensor = np.array(
            image_tensor_list
        )

        print(
            "\nImage tensor shape:",
            image_tensor.shape
        )

        print(
            "Minimum pixel value:",
            image_tensor.min()
        )

        print(
            "Maximum pixel value:",
            image_tensor.max()
        )



# C9. IMAGE AUGMENTATION

from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rotation_range=10,
    width_shift_range=0.05,
    height_shift_range=0.05,
    zoom_range=0.10,
    horizontal_flip=True
)

if chest_path is not None and len(image_tensor_list) > 0:

    sample_batch = image_tensor[
        :4
    ]

    augmented_images = []

    for img in sample_batch:

        img_batch = np.expand_dims(
            img,
            axis=0
        )

        generator = datagen.flow(
            img_batch,
            batch_size=1
        )

        augmented = next(
            generator
        )[0]

        augmented_images.append(
            augmented
        )

    plt.figure(figsize=(12, 6))

    for i, img in enumerate(
        augmented_images
    ):

        plt.subplot(
            1,
            len(augmented_images),
            i + 1
        )

        plt.imshow(
            img.squeeze(),
            cmap="gray"
        )

        plt.title(
            "Augmented"
        )

        plt.axis("off")

    plt.tight_layout()
    plt.show()


# C10. SAVE DE-IDENTIFIED IMAGE COPY
#
# Re-saving the image removes the original metadata.
# This creates a clean image-only copy.

DEID_IMAGE_DIR = os.path.join(
    IMAGE_DIR,
    "deidentified"
)

os.makedirs(
    DEID_IMAGE_DIR,
    exist_ok=True
)

if chest_path is not None:

    for i, file in enumerate(
        valid_images["file"].head(10)
    ):

        try:

            img = Image.open(
                file
            ).convert("L")

            output_file = os.path.join(
                DEID_IMAGE_DIR,
                f"chest_xray_{i}.png"
            )

            img.save(
                output_file
            )

        except:
            pass

    print(
        "\nDe-identified image copies saved to:",
        DEID_IMAGE_DIR
    )



# PART D - SIGNAL DATA
# MIT-BIH ARRHYTHMIA DATABASE


print("\n")
print("=" * 70)
print("PART D - SIGNAL DATA")
print("=" * 70)



# D1. DOWNLOAD MIT-BIH DATABASE

print(
    "\nDownloading MIT-BIH ECG record..."
)

record_name = "100"

try:

    record = wfdb.rdrecord(
        record_name,
        pn_dir="mitdb"
    )

    annotation = wfdb.rdann(
        record_name,
        "atr",
        pn_dir="mitdb"
    )

    print(
        "Record loaded successfully."
    )

except Exception as e:

    print(
        "Could not download MIT-BIH record."
    )

    print(e)

    record = None
    annotation = None



# D2. INSPECT ECG SIGNAL


if record is not None:

    print(
        "\nSampling frequency:",
        record.fs,
        "Hz"
    )

    print(
        "Number of samples:",
        record.sig_len
    )

    print(
        "Number of channels:",
        record.n_sig
    )

    print(
        "Channel names:",
        record.sig_name
    )



# D3. EXTRACT ECG SIGNAL


if record is not None:

    ecg = record.p_signal[:, 0]

    fs = record.fs

    print(
        "\nECG signal shape:",
        ecg.shape
    )



# D4. RAW ECG VISUALIZATION


if record is not None:

    duration_seconds = 10

    samples = int(
        duration_seconds * fs
    )

    time_axis = np.arange(
        samples
    ) / fs

    plt.figure(figsize=(15, 5))

    plt.plot(
        time_axis,
        ecg[:samples]
    )

    plt.title(
        "Raw ECG Signal - MIT-BIH Record 100"
    )

    plt.xlabel(
        "Time (seconds)"
    )

    plt.ylabel(
        "Amplitude (mV)"
    )

    plt.grid()

    plt.show()



# D5. ECG NOISE REMOVAL

def bandpass_filter(
    ecg_signal,
    lowcut=0.5,
    highcut=40,
    fs=360,
    order=4
):

    nyquist = 0.5 * fs

    low = lowcut / nyquist
    high = highcut / nyquist

    b, a = signal.butter(
        order,
        [low, high],
        btype="band"
    )

    filtered = signal.filtfilt(
        b,
        a,
        ecg_signal
    )

    return filtered


if record is not None:

    filtered_ecg = bandpass_filter(
        ecg,
        lowcut=0.5,
        highcut=40,
        fs=fs
    )



# D6. RAW VS FILTERED ECG

if record is not None:

    samples = int(
        10 * fs
    )

    time_axis = np.arange(
        samples
    ) / fs

    plt.figure(figsize=(15, 5))

    plt.plot(
        time_axis,
        ecg[:samples],
        label="Raw ECG"
    )

    plt.plot(
        time_axis,
        filtered_ecg[:samples],
        label="Filtered ECG"
    )

    plt.title(
        "ECG Before and After Noise Removal"
    )

    plt.xlabel(
        "Time (seconds)"
    )

    plt.ylabel(
        "Amplitude"
    )

    plt.legend()

    plt.grid()

    plt.show()



# D7. ECG SIGNAL NORMALIZATION


if record is not None:

    ecg_scaler = StandardScaler()

    normalized_ecg = ecg_scaler.fit_transform(
        filtered_ecg.reshape(-1, 1)
    ).flatten()

    print(
        "Normalized ECG mean:",
        normalized_ecg.mean()
    )

    print(
        "Normalized ECG standard deviation:",
        normalized_ecg.std()
    )



# D8. SIGNAL WINDOWING

WINDOW_SECONDS = 2

if record is not None:

    window_size = int(
        WINDOW_SECONDS * fs
    )

    windows = []

    for start in range(
        0,
        len(normalized_ecg) - window_size,
        window_size
    ):

        window = normalized_ecg[
            start:start + window_size
        ]

        windows.append(
            window
        )

    windows = np.array(
        windows
    )

    print(
        "\nECG window shape:",
        windows.shape
    )



# D9. ECG WINDOW VISUALIZATION


if record is not None and len(windows) > 0:

    plt.figure(figsize=(12, 5))

    plt.plot(
        windows[0]
    )

    plt.title(
        "Example Preprocessed ECG Window"
    )

    plt.xlabel(
        "Sample"
    )

    plt.ylabel(
        "Normalized Amplitude"
    )

    plt.grid()

    plt.show()



# D10. SIGNAL FEATURE EXTRACTION

if record is not None:

    signal_features = []

    for window in windows[:100]:

        features = {

            "mean":
                np.mean(window),

            "std":
                np.std(window),

            "min":
                np.min(window),

            "max":
                np.max(window),

            "range":
                np.ptp(window),

            "median":
                np.median(window),

            "energy":
                np.mean(window ** 2),

            "rms":
                np.sqrt(
                    np.mean(window ** 2)
                ),

            "skewness":
                stats.skew(window),

            "kurtosis":
                stats.kurtosis(window)
        }

        signal_features.append(
            features
        )

    signal_feature_df = pd.DataFrame(
        signal_features
    )

    print(
        "\nExtracted ECG features:"
    )

    display(
        signal_feature_df.head()
    )


# D11. R-PEAK DETECTION

if record is not None:

    distance = int(
        0.25 * fs
    )

    peaks, properties = signal.find_peaks(
        filtered_ecg,
        distance=distance,
        prominence=0.3
    )

    print(
        "\nNumber of detected ECG peaks:",
        len(peaks)
    )



# D12. ECG WITH DETECTED PEAKS

if record is not None:

    samples = int(
        10 * fs
    )

    valid_peaks = peaks[
        peaks < samples
    ]

    time_axis = np.arange(
        samples
    ) / fs

    plt.figure(figsize=(15, 5))

    plt.plot(
        time_axis,
        filtered_ecg[:samples],
        label="Filtered ECG"
    )

    plt.scatter(
        valid_peaks / fs,
        filtered_ecg[
            valid_peaks
        ],
        marker="o",
        label="Detected Peaks"
    )

    plt.title(
        "ECG Signal with Detected Peaks"
    )

    plt.xlabel(
        "Time (seconds)"
    )

    plt.ylabel(
        "Amplitude"
    )

    plt.legend()

    plt.grid()

    plt.show()



# D13. HEART RATE ESTIMATION

if record is not None:

    if len(peaks) > 1:

        rr_intervals = np.diff(
            peaks
        ) / fs

        mean_rr = np.mean(
            rr_intervals
        )

        estimated_heart_rate = (
            60 / mean_rr
        )

        print(
            "Estimated heart rate:",
            round(
                estimated_heart_rate,
                2
            ),
            "BPM"
        )


# FINAL SUMMARY

print("\n")
print("=" * 70)
print("MULTIMODAL MEDICAL DATA LAB COMPLETED")
print("=" * 70)

print("""
A. TABULAR DATA
----------------
Dataset:
Pima Indians Diabetes Dataset

Completed:
✓ Dataset acquisition
✓ Dataset inspection
✓ Duplicate removal
✓ Healthcare-specific missing value detection
✓ Invalid zero handling
✓ Median imputation
✓ Clinical value validation
✓ Feature engineering
✓ EDA
✓ Outlier detection
✓ Hypothesis testing
✓ Chi-square test
✓ ANOVA
✓ Feature selection
✓ Feature scaling
✓ Class imbalance handling


B. TEXTUAL DATA
----------------
Dataset:
MTSamples Medical Transcriptions

Completed:
✓ Dataset acquisition
✓ Text inspection
✓ Duplicate removal
✓ PHI-related pattern removal
✓ Text cleaning
✓ Tokenization
✓ Vocabulary creation
✓ Token sequences
✓ Text length features
✓ Medical specialty analysis
✓ EDA


C. IMAGE DATA
----------------
Dataset:
Chest X-Ray Images (Pneumonia)

Completed:
✓ Dataset acquisition
✓ Image inspection
✓ Class distribution
✓ Sample visualization
✓ Image-size analysis
✓ Metadata/EXIF inspection
✓ Metadata-safe re-saving
✓ Image resizing
✓ Pixel normalization
✓ Tensor preparation
✓ Data augmentation


D. SIGNAL DATA
----------------
Dataset:
MIT-BIH Arrhythmia Database

Completed:
✓ ECG acquisition
✓ Signal inspection
✓ Raw ECG visualization
✓ Bandpass noise filtering
✓ Signal normalization
✓ Fixed-size windowing
✓ Signal feature extraction
✓ R-peak detection
✓ Heart-rate estimation
✓ Preprocessed ECG visualization


Healthcare-specific considerations:
✓ Missing values can be hidden as valid-looking zeros
✓ Medical data may contain PHI
✓ Image metadata can contain patient information
✓ ECG contains physiological and acquisition noise
✓ Class imbalance can bias models
✓ Clinical values require plausibility checks
""")
```
