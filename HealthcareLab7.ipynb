```python
# 1. INSTALL / IMPORT

!pip -q install tensorflow scikit-learn

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("TensorFlow :", tf.__version__)
print("Scikit-learn :", __import__("sklearn").__version__)



# 2. LOAD PIMA INDIANS DIABETES DATASET

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

df = pd.read_csv(
    url,
    names=columns
)

print("\nDataset shape:", df.shape)

display(df.head())



# 3. DATA EXPLORATION


print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())



# 4. DATA PREPROCESSING


zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

print("\nZero values before treatment:")

for col in zero_columns:
    print(
        col,
        ":",
        (df[col] == 0).sum()
    )


# Replace zeros with NaN
df[zero_columns] = df[
    zero_columns
].replace(
    0,
    np.nan
)


# Replace missing values with median
for col in zero_columns:

    df[col] = df[col].fillna(
        df[col].median()
    )


print("\nMissing values after preprocessing:")

print(df.isnull().sum())



# 5. FEATURES AND TARGET


X = df.drop(
    "Outcome",
    axis=1
)

y = df["Outcome"]


print("\nFeatures:", X.shape)
print("Target :", y.shape)

print("\nClass distribution:")
print(y.value_counts())


# 6. TRAIN TEST SPLIT


X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\nTraining samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])


# 7. FEATURE SCALING

scaler = StandardScaler()

X_train = scaler.fit_transform(
    X_train
)

X_test = scaler.transform(
    X_test
)


# Convert to numpy
X_train = np.asarray(X_train).astype(
    np.float32
)

X_test = np.asarray(X_test).astype(
    np.float32
)

y_train = np.asarray(y_train).astype(
    np.int32
)

y_test = np.asarray(y_test).astype(
    np.int32
)


# 8. VISUALIZATION

plt.figure(figsize=(7,5))

sns.countplot(
    x=df["Outcome"]
)

plt.title(
    "Pima Indians Diabetes Class Distribution"
)

plt.xlabel("Outcome")
plt.ylabel("Number of Patients")

plt.show()


# 9. CORRELATION HEATMAP

plt.figure(figsize=(10,7))

sns.heatmap(
    df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Feature Correlation Heatmap"
)

plt.show()



# 10. CUSTOM KERAS + SCIKIT-LEARN CLASSIFIER


class KerasGridSearchClassifier(
    BaseEstimator,
    ClassifierMixin
):

    def __init__(
        self,
        learning_rate=0.001,
        init_mode="uniform",
        activation="relu",
        dropout_rate=0.0,
        neurons=16,
        batch_size=32,
        epochs=50
    ):

        self.learning_rate = learning_rate
        self.init_mode = init_mode
        self.activation = activation
        self.dropout_rate = dropout_rate
        self.neurons = neurons
        self.batch_size = batch_size
        self.epochs = epochs

        self.model_ = None



    # CREATE KERAS MODEL


    def _create_model(self):

        model = Sequential()

        # Input layer
        model.add(
            Input(
                shape=(
                    self.n_features_,
                )
            )
        )

        # Hidden layer
        model.add(
            Dense(
                self.neurons,

                activation=self.activation,

                kernel_initializer=self.init_mode
            )
        )

        # Dropout
        if self.dropout_rate > 0:

            model.add(
                Dropout(
                    self.dropout_rate
                )
            )


        # Output layer
        model.add(
            Dense(
                1,

                activation="sigmoid",

                kernel_initializer=self.init_mode
            )
        )


        # Optimizer
        optimizer = Adam(
            learning_rate=self.learning_rate
        )


        # Compile
        model.compile(

            optimizer=optimizer,

            loss="binary_crossentropy",

            metrics=["accuracy"]
        )


        return model



    # FIT


    def fit(
        self,
        X,
        y
    ):

        # Number of input features
        self.n_features_ = X.shape[1]

        # Create model
        self.model_ = self._create_model()

        # Train
        self.model_.fit(

            X,
            y,

            batch_size=self.batch_size,

            epochs=self.epochs,

            verbose=0
        )

        # Classes required by sklearn
        self.classes_ = np.unique(y)

        return self



    # PREDICT PROBABILITY

    def predict_proba(
        self,
        X
    ):

        probabilities = self.model_.predict(
            X,
            verbose=0
        ).ravel()


        return np.column_stack(
            [
                1 - probabilities,
                probabilities
            ]
        )



    # PREDICT


    def predict(
        self,
        X
    ):

        probabilities = self.predict_proba(
            X
        )[:, 1]


        return (
            probabilities >= 0.5
        ).astype(int)



# 11. CREATE MODEL


model = KerasGridSearchClassifier()

print("\nCustom Keras classifier created successfully.")



# 12. GRID SEARCH PARAMETERS


param_grid = {


    # Tune batch size

    "batch_size": [
        16,
        32
    ],


    # Tune training epochs

    "epochs": [
        30
    ],



    # Tune learning rate

    "learning_rate": [
        0.001,
        0.01
    ],



    # Tune network weight initialization

    "init_mode": [
        "uniform",
        "he_uniform"
    ],



    # Tune activation function

    "activation": [
        "relu",
        "tanh"
    ],



    # Tune dropout

    "dropout_rate": [
        0.0,
        0.2
    ],


    # Tune hidden-layer neurons

    "neurons": [
        16,
        32
    ]
}



# 13. CALCULATE GRID SIZE

number_of_combinations = 1

for values in param_grid.values():

    number_of_combinations *= len(
        values
    )


print("\n" + "=" * 60)

print(
    "Total parameter combinations:",
    number_of_combinations
)

print(
    "CV folds:",
    3
)

print(
    "Total approximate model fits:",
    number_of_combinations * 3
)

print("=" * 60)



# 14. GRID SEARCH


grid = GridSearchCV(

    estimator=model,

    param_grid=param_grid,

    scoring="accuracy",

    cv=3,

    n_jobs=1,

    verbose=2,

    return_train_score=True
)


print("\nStarting Grid Search...")
print("Please wait...")


grid_result = grid.fit(

    X_train,

    y_train
)


print("\nGrid Search completed successfully!")



# 15. BEST PARAMETERS

print("\n" + "=" * 60)
print("BEST PARAMETERS")
print("=" * 60)

for parameter, value in (
    grid_result.best_params_.items()
):

    print(
        f"{parameter}: {value}"
    )


print(
    "\nBest Cross-Validation Accuracy:",
    round(
        grid_result.best_score_,
        4
    )
)



# 16. GRID SEARCH RESULTS


results = pd.DataFrame(
    grid_result.cv_results_
)

results = results.sort_values(
    by="mean_test_score",
    ascending=False
)


print("\nTop 10 parameter combinations:")

display(

    results[
        [
            "mean_test_score",
            "std_test_score",
            "param_batch_size",
            "param_epochs",
            "param_learning_rate",
            "param_init_mode",
            "param_activation",
            "param_dropout_rate",
            "param_neurons"
        ]
    ].head(10)

)


# 17. BEST MODEL


best_model = grid_result.best_estimator_



# 18. TEST SET PREDICTION


y_pred = best_model.predict(
    X_test
)


# 19. EVALUATION

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)


print("\n" + "=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)

print(
    f"Accuracy  : {accuracy:.4f}"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall    : {recall:.4f}"
)

print(
    f"F1 Score  : {f1:.4f}"
)


# 20. CLASSIFICATION REPORT

print("\nClassification Report:")

print(

    classification_report(

        y_test,

        y_pred,

        target_names=[
            "No Diabetes",
            "Diabetes"
        ],

        zero_division=0
    )

)


# 21. CONFUSION MATRIX

cm = confusion_matrix(
    y_test,
    y_pred
)


plt.figure(
    figsize=(6,5)
)

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "No Diabetes",
        "Diabetes"
    ],

    yticklabels=[
        "No Diabetes",
        "Diabetes"
    ]
)


plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "Confusion Matrix - Best Grid Search Model"
)

plt.show()


# 22. PERFORMANCE SUMMARY

summary = pd.DataFrame({

    "Metric": [

        "Best CV Accuracy",

        "Test Accuracy",

        "Test Precision",

        "Test Recall",

        "Test F1 Score"

    ],

    "Score": [

        grid_result.best_score_,

        accuracy,

        precision,

        recall,

        f1

    ]

})


display(summary)


# 23. SAVE BEST KERAS MODEL

best_model.model_.save(
    "pima_diabetes_best_model.keras"
)

print(
    "\nBest model saved as:"
)

print(
    "pima_diabetes_best_model.keras"
)
```
