```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import zipfile
import os

columns = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal",
    "target"
]

zip_file_path = "/content/heart+disease.zip"
extract_dir = "/content/extracted_data"
csv_file_in_zip = "processed.cleveland.data"

try:
    # Create directory for extraction if it doesn't exist
    os.makedirs(extract_dir, exist_ok=True)

    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extract(csv_file_in_zip, extract_dir)

    csv_path = os.path.join(extract_dir, csv_file_in_zip)
    df = pd.read_csv(
        csv_path,
        names=columns,
        na_values="?"
    )
    print(df)

except zipfile.BadZipFile:
    print(f"Error: The file '{zip_file_path}' is not a valid ZIP file. Please check the file integrity or re-upload it.")
except FileNotFoundError:
    print(f"Error: The file '{zip_file_path}' was not found. Please ensure it's uploaded correctly.")
except KeyError:
    print(f"Error: The file '{csv_file_in_zip}' was not found inside '{zip_file_path}'.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
